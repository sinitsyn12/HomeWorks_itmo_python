## 2. Пропущенные значения

## 2.1. Найдите пропуски

```python
## кол-во пропусков и доля

missing_data = pd.DataFrame({
    'Количество пропусков': df.isna().sum(),
    'Доля пропусков в %': (df.isna().mean() * 100).round(2)
})

missing_data
```

```python
# # вывод только столбцов с пропусками по убыванию

missing_data = missing_data[
    missing_data['Количество пропусков'] > 0
].sort_values(
    'Количество пропусков',
    ascending=False
)

missing_data
```

Больше всего пропусков находится в license. Также много пропусков в `price`, `last_review` и `reviews_per_month`.

Пропуски имеют разную критичность. Например, отсутствие price мешает анализировать стоимость жилья. Пропуск в license менее критичен для анализа цен и типа жилья.

## 2.2. Исследуйте причину пропусков

```python
df.loc[df['reviews_per_month'].isna(),'number_of_reviews'].value_counts()
```

```python
# Проверяем отзывы в месяц у объявлений без отзывов

df.loc[df['number_of_reviews'] == 0, 'reviews_per_month'].isna().value_counts()
```

Все 8 616 пропусков в `reviews_per_month` находятся у объявлений с нулевым количеством отзывов. И наоборот: у всех объявлений без отзывов значение `reviews_per_month` пропущено.

Следовательно, пропуск здесь означает не неизвестное значение, а отсутствие отзывов. Такие пропуски можно будет обоснованно заменить на 0.

## 2.3. Выберите стратегию обработки

```python
## будем работать в копии чтобы не менять исходный дф

df_clean = df.copy()
```

```python
## меняем пропущенные значения отзывов в месяц на 0

df_clean['reviews_per_month'] = (df_clean['reviews_per_month'].fillna(0))
```

```python
## меняем отсутвие лицензии на Not stated (Ns)

df_clean['license'] = (df_clean['license'].fillna('Ns'))
```

```python
df_clean[['reviews_per_month', 'license']].isna().sum()
```

В `reviews_per_month` пропуски заменены на 0, поскольку проверка в задание выше показала, что они соответствуют объявлениям без отзывов.

В `license` пропуски заменены на категорию «Ns». Это позволяет явно зафиксировать, что лицензия не указана.

```python
## визуализируем

fig = px.box(
    price_data,
    y='price',
    points='outliers',
    title='Поиск потенциальных выбросов в цене',
    labels={
        'price': 'Цена за ночь'
    }
)

fig.show()
```

```python
## метод IQR

q1 = price_data['price'].quantile(0.25)
q3 = price_data['price'].quantile(0.75)

iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

print('Первый квартиль:', q1)
print('Третий квартиль:', q3)
print('IQR:', iqr)
print('Нижняя граница:', lower_limit)
print('Верхняя граница:', upper_limit)
```

```python
price_outliers = price_data[
    (price_data['price'] < lower_limit) |
    (price_data['price'] > upper_limit)
]

outliers_count = len(price_outliers)

outliers_share_all = (
    outliers_count / len(df) * 100
)

outliers_share_known = (
    outliers_count / len(price_data) * 100
)

print('Количество значений за пределами границ:', outliers_count)
print(
    'Доля от всего датасета:',
    round(outliers_share_all, 2),
    '%'
)
print(
    'Доля среди объявлений с известной ценой:',
    round(outliers_share_known, 2),
    '%'
)
```
