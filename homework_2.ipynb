# Задание №1. NumPy

Данны массивы:

*   prices = np.array([1200, 800, 2500, 3100, 450, 1800])
*   counts = np.array([2, 1, 3, 1, 5, 2])

Выполните:
1. получите массив стоимости каждой позиции: price * count;
2. найдите общую стоимость всех позиций;
3. найдите среднюю цену;
4. получите только цены больше 1500;
5. получите цены от 1000 до 3000 включительно;
6. посчитайте количество цен больше 1500.

```python
import numpy as np

prices = np.array([1200, 800, 2500, 3100, 450, 1800])
counts = np.array([2, 1, 3, 1, 5, 2])
```

```python
## получите массив стоимости каждой позиции: price * count;
print(prices * counts)
```

```python
## найдите общую стоимость всех позиций;
print(sum(prices * counts))
```

```python
## найдите среднюю цену;
print(np.mean(prices))
```

```python
## получите только цены больше 1500;
print(prices[prices >=1500])
```

```python
## получите цены от 1000 до 3000 включительно;
print(prices[(prices >= 1000) & (prices <= 3000)])
```

```python
## посчитайте количество цен больше 1500.
print(len(prices[prices > 1500]))
```
