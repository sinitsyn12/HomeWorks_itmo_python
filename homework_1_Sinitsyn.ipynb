# Задание №2 Подготовка данных к анализу

Нужно обработать исходный список, убрав лишние прообелы, некоректные данные и т.д.  

*Исходный список менять не нужно. Копируем в переменую clean_orders*

**Желаемый формат данных**
* customer — строка без лишних пробелов по краям;
* city — строка без лишних пробелов и в нижнем регистре;
* amount — число типа float, если исходное значение можно преобразовать в число;
* если сумму преобразовать нельзя, amount должен быть равен None;
* остальные поля сохраняются.

```python
## сразу сделаем функцию обработки amount, чтобы применить её в функции ниже. Вернувшись с 4-го задания я добавил ещё except для TypError.

def clean_amount(amount):
  try:
    return float(amount)
  except (ValueError, TypeError):
    return None
```

```python
## аналогично почистим сити

def clean_city(city):
  return city.lower().strip()
```

```python
## Создаем копию нашего списка и чистим его от шума + добавляем туда новый amount и city

def clean_order (order):
  new_order = order.copy()

  new_order["customer"] = new_order["customer"].strip()
  new_order["city"] = clean_city(new_order["city"])
  new_order["amount"] = clean_amount(new_order["amount"])

  return new_order
```

```python
## Задаём новую переменную, куда будем складывать очищенную элементы списка

clean_orders = []

for order in orders:
  func_order = clean_order(order)
  clean_orders.append(func_order)

print(clean_orders)
```

```python
## Добавил вывод для себя, чтобы удобнее было проверять

for order in clean_orders:
  print(order)
```

# Задание № 3 Проанализируйте очищенные данные

**Нужно получить следующее:**

1. количество заказов со статусом paid;
2. количество заказов со статусом cancelled;
3. общую сумму оплаченных заказов, для которых известна корректная сумма;
4. среднюю сумму оплаченного заказа среди заказов с корректной суммой;
5. множество уникальных городов;
6. словарь с количеством заказов по каждому городу;
7. список имён клиентов из Москвы, у которых:
  
  *   статус заказа paid;
  *   сумма заказа больше 1000.






## Подзадания 1 и 2

```python
## Вводим переменные счётчики для заказов с разным статусом. Через цикл считаем кол-во и выводим результат.

paid_total = 0
cancelled_total = 0

for order in clean_orders:
  if order['status'] == "paid":
    paid_total += 1

  elif order['status'] == "cancelled":
    cancelled_total += 1

print(f" Кол-во заказов со статусом paid: {paid_total}\n Кол-во заказов со статусом cancelled: {cancelled_total}")
```

## Подзадания 3 и 4

```python
## Задал пустой список для нужных значений. Циклом прошёлся по исходному списку и отобрал только оплаченные и только численные значения.

paid_orders_total = []

for order in clean_orders:
  if type(order['amount']) == float and order['status'] == "paid":
    paid_orders_total.append(order['amount'])

print(paid_orders_total)
```

```python
## Выведем нужные значения: суммы и среднего

paid_orders_mean = sum(paid_orders_total) / len(paid_orders_total)

print(f'Сумма всех оплаченных заказов: {sum(paid_orders_total)}\nСреднее всех оплаченных заказов: {paid_orders_mean}')
```

## Подзадания 6 и 7

```python
## количество заказов по городам

city_count = {}

for order in clean_orders:
    city = order['city']
    if city in city_count:
        city_count[city] += 1
    else:
        city_count[city] = 1

print(city_count)
```

```python
## Cписок имен клиентов из москвы по условию

moscow_clients = []

for order in clean_orders:
    if order['city'] == 'москва' and order['status'] == 'paid' and order['amount'] is not None and order['amount'] > 1000:
        moscow_clients.append(order['customer'])

print(moscow_clients)
```
