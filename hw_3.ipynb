# Задание 2. Преобразования столбцов

## 2.1. Размер семьи

Создайте столбец family_size, равный:  
SibSp + Parch + 1

```python
## добавляем 1 потому что считаем сразу с человеком в этой строке. Его сиблинги/родители + он сам.

titanic['family_size'] = titanic['SibSp'] + titanic['Parch'] + 1

titanic.head()
```

## 2.2. Стоимость билета на одного члена семьи

Создайте столбец **fare_per_person**, содержащий стоимость билета на одного члена семьи.

```python
titanic['fare_per_person'] = titanic['Fare'] / titanic['family_size']

titanic.head()
```

## 2.3. Работа со строковыми данными

Используя столбец Name:

* создайте name_length — длину имени;
* создайте логический столбец has_mrs, показывающий, содержит ли имя "Mrs.";
* посчитайте количество строк, где has_mrs == True.

```python
titanic['name_length'] = titanic['Name'].str.len()
titanic['has_mrs'] = titanic['Name'].str.contains('Mrs.')

titanic.head()
```

```python
## количество True можно посчитать через метод sum, так как он пропускает False
print(titanic['has_mrs'].sum())
```

## 4.1. Исследуйте распределение пассажиров по классам (Pclass).

В нашем датасете Pclass — это категория: пассажир относится к 1, 2 или 3 классу. Здесь лучше всего использовать столбачутую диаграмму, так мы наглядно сможем увидеть распределение.

```python
class_counts = titanic['Pclass'].value_counts()

fig = px.bar(
    x=class_counts.index,
    y=class_counts.values,
    labels={
        'x': 'Класс пассажира',
        'y': 'Количество пассажиров'
    },
    title = 'Распределение пассажиров по классам'
)

fig.show()
```

По графику мы явно видиим, что самая многочисленная группа пассажиров 3-го класса: почти 500 человек. За ними идёт 1-й класс: чуть более 200 и самый малочисленный 2-ой класс: 184 человека.

## 4.2 Исследуйте распределение возраста пассажиров (Age).

Для переменной возраста у нас множество значений, поэтому столбики уже не подойдут. Здесь использую гистограмму, так как она показывает как распределены пассажиры по возрасту. Так, мы можем с первых секунд понять где пассажиров больше, где меньше, прикинуть средний возраст.

```python
fig = px.histogram(
    titanic,
    x='Age',
    nbins=20,
    title='Распределение пассажиров по возрасту',
    labels={'Age': 'Возраст пассажира'}
)

fig.update_layout(
    yaxis_title='Количество пассажиров'
)

fig.show()
```

Тут мы видим, что большинство пассажиров лайнера в возрасте от ~18 до 32. Начиная с 40 лет распределение уменьшается.Пассажиров страше 73 лет — 1 человек.

# 4.3. Исследуйте распределение стоимости билета (Fare).  
Обратите внимание не только на типичные значения, но и на разброс и крайние наблюдения.


Стоимость билета такая же подвижная величина, как и возраст. Поэтому оставляем гистограмму, с добавлением разброс через "box".

```python
fig = px.histogram(
    titanic,
    x='Fare',
    nbins=30,
    marginal='box',
    title='Распределение стоимости билетов',
    labels={'Fare': 'Стоимость билета'}
)

fig.update_layout(
    yaxis_title='Количество пассажиров'
)

fig.show()
```

Большинство пассажиров купили билеты до 100 фунтов (титаник же из Англии плыл?). Кто-то один купил билет за 500 фунтов.

## 4.4 Сравните стоимость билета (Fare) между классами пассажиров (Pclass).

Здесь будем использовать boxplot, так как сравниваем числовую переменную между несколькими группами: стоимость и класс пассажира.

```python
fig = px.box(
    titanic,
    x='Pclass',
    y='Fare',
    title='Стоимость билетов по классам пассажиров',
    labels={
        'Pclass': 'Класс пассажира',
        'Fare': 'Стоимость билета'
    }
)

fig.show()
```

стоимость билетов сильно разниться взависимости от класса. В первом классе большой разброс цен, в третьем чуть меньше, без явных выборосов.

## 4.5 Исследуйте связь между возрастом (Age) и стоимостью билета (Fare).

На одной визуализации должно быть видно, какие пассажиры выжили, а какие нет.

Будем использовать scatter, так как нам нужно установить связь двух величин + визуально разделить их на две группы: выжил или умер.

```python
fig = px.scatter(
    titanic,
    x='Age',
    y='Fare',
    color='Survived',
    title='Связь возраста и стоимости билета с выживаемостью',
    labels={
        'Age': 'Возраст пассажира',
        'Fare': 'Стоимость билета',
        'Survived': 'Выживаемость'
    }
)

fig.show()
```

Прямые выводы из графика сделать сложно. Визуально кажется, что выживших пассажиров с более дорогими билетами больше, чем аналогичных с более дешевыми. Но явно график нам об этом не говорит. По возрасту четкого разделения нет, нужно исследовать дальше

## 4.6 Сравните стоимость билета (Fare) между классами пассажиров (Pclass).

В этом случае обе переменных категориальные, поэтому будем использовать столбчатую диаграмму с доп разделением через color

```python
titanic.head()
```

```python
fig = px.histogram(
    titanic,
    x='Pclass',
    color='Survived',
    barmode='group',
    title='Выживаемость пассажиров в разных классах',
    labels={
        'Pclass': 'Класс пассажира',
        'Survived': 'Выживаемость'
    }
)

fig.update_layout(
    yaxis_title='Количество пассажиров'
)

fig.show()
```

## 4.8. Корреляци
Для признаков:

* Age
* Fare
* SibSp
* Parch
* family_size

посчитайте корреляции;
визуализируйте их;
укажите наиболее заметную положительную связь.

```python
## получаем матрицу корреляции

corr = titanic[['Age', 'Fare', 'SibSp', 'Parch', 'family_size']].corr()

corr
```

```python
## строим визуализацию матрицы кореляяции. text_auto показывает нам значение прямо внутри ячейки с округлением до 2-х знаков.

fig = px.imshow(
    corr,
    text_auto='.2f',
    title='Корреляция числовых признаков',
    labels={
        'color': 'Корреляция'
    },
    zmin=-1,
    zmax=1
)

fig.show()
```

Наиболее заметная положительная связь наблюдается между SibSp и family_size. Это ожидаемо, поскольку SibSp входит в формулу расчёта размера семьи: family_size = SibSp + Parch + 1
