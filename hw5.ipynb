# 4. loc: условное преобразование

Создайте столбец:

* stay_type  

По значению minimum_nights:

* меньше 7 → "short";
* от 7 до 30 включительно → "medium";
* больше 30 → "long".
* Сделайте преобразование через loc.

После этого:

1. покажите value_counts();
2. проверьте несколько строк каждой категории;
3. объясните, почему здесь loc удобнее, чем изменение таблицы в цикле.

```python
## делаем пустой столбец

df_work["stay_type"] = ""
```

```python
df_work.loc[df_work["minimum_nights"] < 7, "stay_type"] = "short"
```

```python
df_work.loc[
    (df_work["minimum_nights"] >= 7)
    & (df_work["minimum_nights"] <= 30),
    "stay_type"
] = "medium"
```

```python
df_work.loc[df_work["minimum_nights"] > 30, "stay_type"] = "long"
```

```python
## проверим что мы натворили создав тестовый датафрейм

stay_examples = df_work.groupby(
    "stay_type"
).head(3)

stay_examples[["minimum_nights", "stay_type"]].sort_values("stay_type")
```

Почему loc удобнее в этом случае?  
Метод loc позволяет выбрать сразу все строки, которые соответствуют определённому условию, и записать в них новое значение. Поэтому не требуется перебирать таблицу построчно в цикле.

# 5. apply и lambda

## 5.1. Функция по нескольким столбцам

Создайте признак: eview_activity

Правила:

1. если number_of_reviews == 0 → "no_reviews";
2. иначе если reviews_per_month >= 2 → "active";
3. иначе → "rare".

Логика зависит сразу от двух столбцов, поэтому используйте отдельную функцию и: df_work.apply(..., axis=1)

Если reviews_per_month пропущен, но number_of_reviews > 0, отнесите объект к "rare".

```python
## создаем функцию

def get_review_activity(row):
    if row["number_of_reviews"] == 0:
        return "no_reviews"
    elif pd.isna(row["reviews_per_month"]):
        return "rare"
    elif row["reviews_per_month"] >= 2:
        return "active"
    else:
        return "rare"
```

```python
## применяем функцию

df_work["review_activity"] = df_work.apply(get_review_activity,axis=1)
```

```python
df_work[
    ["number_of_reviews",
     "reviews_per_month",
     "review_activity"]
].head()
```

## 5.2. lambda

Создайте ещё один небольшой признак через lambda.

Он должен быть осмысленным и использоваться только один раз.

В Markdown напишите, почему для этой задачи короткая lambda была уместна.

Создадим ещё один признак is_available_180_plus:
* 1 — жильё доступно не менее 180 дней в году;
* 0 — жильё доступно меньше 180 дней.

```python
df_work["is_available_180_plus"] = df_work["availability_365"].apply(lambda days: 1 if days >= 180 else 0)
```

```python
df_work["is_available_180_plus"].value_counts()
```

lambda удобна, потому что условие короткое и применяется только один раз, поэтому отдельная именованная функция здесь не требуется.

# 6. Кодирование номинальной категории

Преобразуйте room_type с помощью One-Hot Encoding.  

Требования:

* создайте dummy-столбцы;
* добавьте их к df_work или создайте отдельную таблицу;
* покажите первые строки;
* объясните, почему One-Hot Encoding здесь лучше простого присвоения чисел 0, 1, 2, ....


```python
room_type_dummies = pd.get_dummies(
    df_work["room_type"],
    prefix="room_type",
    dtype=int
)
```

```python
room_type_dummies.head()
```

One-Hot Encoding создаёт отдельный бинарный столбец для каждого типа жилья. Этот способ лучше простого присвоения чисел 0, 1, 2, ..., потому что между типами жилья нет естественного порядка. Числовые коды могли бы создать для модели ложную зависимость, например, что один тип жилья больше другого. Ситуация аналогичная сокращениям районов.

Выберите признаки:

* price
* minimum_nights
* number_of_reviews
* availability_365

```python
numeric_columns = [
    "price",
    "minimum_nights",
    "number_of_reviews",
    "availability_365"
]

numeric_data = df_work[numeric_columns].copy()
```

## 7.1. StandardScaler

Создайте копию выбранных признаков и примените StandardScaler.

Сравните describe() до и после масштабирования.

Ответьте:

* что произошло со средними значениями;
* что произошло со стандартными отклонениями;
* сохранился ли порядок объектов внутри каждого признака.

```python
from sklearn.preprocessing import StandardScaler
```

```python
standard_scaler = StandardScaler()
```

```python
## создаём копию и применяем масштабирование

df_standard = numeric_data.copy()

df_standard[numeric_columns] = standard_scaler.fit_transform(numeric_data)
```

```python
## проверяем
df_standard.describe().loc[["mean", "std", "min", "max"]].round(2)
```

После применения StandardScaler средние значения признаков стали примерно равны нулю, а стандартные отклонения — единице. Масштаб значений изменился, но порядок объектов сохранился.

## 7.2. MinMaxScaler

Для тех же признаков отдельно примените MinMaxScaler.

Проверьте минимальные и максимальные значения после преобразования

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
minmax_scaler = MinMaxScaler()
```

```python
df_minmax = numeric_data.copy()

df_minmax[numeric_columns] = minmax_scaler.fit_transform(numeric_data)
```

```python
## проверяем результат

df_minmax.describe().loc[["min", "max"]].round(2)
```

## 7.3. Сравнение

StandardScaler преобразует данные так, чтобы среднее значение каждого признака было примерно равно нулю, а стандартное отклонение — единице. MinMaxScaler переносит значения в диапазон от 0 до 1. Оба способа сохраняют порядок объектов внутри каждого столбца.

```python
## выполняем кодирование для районов

borough_dummies = pd.get_dummies(
    df_work["borough_code"],
    prefix="borough",
    dtype=int
)
```

```python
## кодирование для категории длительности проживания

stay_type_dummies = pd.get_dummies(
    df_work["stay_type"],
    prefix="stay",
    dtype=int)
```
